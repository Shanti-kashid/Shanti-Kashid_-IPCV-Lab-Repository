Sobel, Laplacian, Canny, and Laplacian of Gaussian (LoG) are edge detection filters used to detect object boundaries in images, while the Gaussian filter is a smoothing filter used to reduce noise before edge detection.

**Sobel Filter**

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Apply Sobel Filter in X and Y directions
# ksize=3 indicates a 3x3 kernel
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Combine the two gradients
sobel_combined = cv2.magnitude(sobel_x, sobel_y)

# Convert back to 8-bit unsigned integer for display
sobel_final = np.uint8(sobel_combined)
final_img=cv2.resize(sobel_final, (400, 400))
cv2_imshow(final_img)
cv2.waitKey(0)
```

**Canny Filter**

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load the image
image = img

# Resize the image before processing
resized_for_canny = cv2.resize(image, (400, 400)) # Resize to 400x400 pixels (width, height)

# 2. Apply Gaussian Blur to reduce noise
blurred_image = cv2.GaussianBlur(resized_for_canny, (5, 5), 0)

# 3. Apply Canny Edge Detection
# Thresholds: 50 (min) and 150 (max) - adjust these based on image
edges = cv2.Canny(blurred_image, 50, 150)

# 4. Display the original resized image and edge-detected image
cv2_imshow(resized_for_canny)
cv2_imshow(edges)
```

**Laplassian Filter**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread('/content/drive/My Drive/IPCV/closeup-shot-beautiful-butterfly-with-interesting-textures-orange-petaled-flower.jpg', 0)

# Resize image
img = cv2.resize(img, (400,400))

# Display original resized image
plt.imshow(img, cmap='gray')
plt.title("Resized Original Image")
plt.axis("off")
plt.show()

# Apply Laplacian Filter
laplacian = cv2.Laplacian(img, cv2.CV_64F)

# Convert to absolute values
laplacian = np.uint8(np.absolute(laplacian))
print("Laplacian Edge Image")
cv2_imshow(laplacian)
```

**Log Filter**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('/content/drive/My Drive/IPCV/closeup-shot-beautiful-butterfly-with-interesting-textures-orange-petaled-flower.jpg', 0)

# Display original image
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()

# Gaussian Blur
blur = cv2.GaussianBlur(img,(5,5),0)

# Laplacian after Gaussian
log = cv2.Laplacian(blur, cv2.CV_64F)
log = np.uint8(np.absolute(log))

plt.imshow(log, cmap='gray')
plt.title("LoG Filter")
plt.axis("off")
plt.show()
```
